```python
# Display controller UI
import os
import ipywidgets as widgets
from robot_controller import joint_controller, robot_steering, ee_controller, perception_controller
from IPython.display import display, clear_output

robots = [
    "stretch",
    # "pr2",
    # "hsrb",
    # "tiago_dual",
]

robot_controllers = []

for robot in robots:
    sections = []
    # Base drive + Cartesian IK end-effector control + perception (Stretch only).
    if robot == "stretch":
        sections.append(widgets.HTML("<b>Base drive</b>"))
        sections.append(robot_steering(robot))
        sections.append(widgets.HTML("<b>End-effector (IK) control</b>"))
        sections.append(ee_controller(robot))
        sections.append(widgets.HTML("<b>Perception</b>"))
        sections.append(perception_controller(robot))
        sections.append(widgets.HTML("<b>Joint control</b>"))
    sections.append(
        joint_controller(
            urdf_path=f"{os.getcwd()}/../usd/{robot}/{robot}.urdf",
            prefix=robot,
        )
    )
    robot_controllers.append(widgets.VBox(sections))

tabs = widgets.Tab(children=robot_controllers)
for i in range(len(robots)):
    tabs.set_title(i, robots[i])

clear_output()
display(tabs)
```
